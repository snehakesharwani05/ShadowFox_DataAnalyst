```python
order_id_counts = df["Order ID"].value_counts()

print("Orders appearing more than once:",
      (order_id_counts > 1).sum())

print("\nMaximum appearances of one Order ID:",
      order_id_counts.max())

order_id_counts[order_id_counts > 1].head(20)
```

```python
repeated_ids = order_id_counts[order_id_counts > 1].index

repeated_orders = df[df["Order ID"].isin(repeated_ids)]

repeated_orders.sort_values("Order ID").head(20)
```

```python
id_variation = (
    df.groupby("Order ID")
      .agg(
          rows=("Order ID", "size"),
          unique_status=("Order Status", "nunique"),
          unique_amount=("Order Amount (INR)", "nunique"),
          unique_restaurants=("res_id", "nunique"),
          unique_dates=("Order Date", "nunique")
      )
)

id_variation[id_variation["rows"] > 1].head(20)
```

```python
# Check whether every repeated Order ID has identical records

duplicate_id_check = (
    df.groupby("Order ID")
      .nunique()
)

repeated_id_check = duplicate_id_check.loc[
    df["Order ID"].value_counts()
    .loc[lambda x: x > 1]
    .index
]

# Find repeated Order IDs where any column has different values
inconsistent_ids = repeated_id_check[
    (repeated_id_check > 1).any(axis=1)
]

print("Repeated Order IDs:", len(repeated_id_check))
print("Repeated IDs with different information:", len(inconsistent_ids))

inconsistent_ids
```

```python
df_clean = df.drop_duplicates().copy()

print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
print("Removed rows:", len(df) - len(df_clean))
```

```python
print("Remaining duplicates:", df_clean.duplicated().sum())
```

```python
df_clean["type"] = df_clean["type"].fillna("Unknown")

df_clean["cuisines"] = df_clean["cuisines"].fillna("Unknown")

df_clean["timings"] = df_clean["timings"].fillna("Not Available")

df_clean["address"] = df_clean["address"].fillna("Address Not Available")

df_clean["opentable_support"] = df_clean["opentable_support"].fillna(0)
```

```python
df_clean.isnull().sum()[df_clean.isnull().sum() > 0]
```

```python
df_clean["Month"] = df_clean["Order Date"].dt.month_name()
```

```python
df_clean["Year"] = df_clean["Order Date"].dt.year
```

```python
df_clean["Day"] = df_clean["Order Date"].dt.day_name()
```

```python
df_clean["Quarter"] = df_clean["Order Date"].dt.quarter
```

```python
df_clean["Is_Delivered"] = (
    df_clean["Order Status"] == "Delivered"
).astype(int)
```

```python
df_clean["Is_Cancelled"] = (
    df_clean["Order Status"] == "Cancelled"
).astype(int)
```

```python
df_clean["Has_Promo"] = (
    df_clean["Promo Code Applied"] == "Yes"
).astype(int)
```

```python
df_clean["Free_Delivery"] = (
    df_clean["Has Free Delivery"] == "Yes"
).astype(int)
```

```python
df_clean["Completed_Revenue"] = np.where(
    df_clean["Order Status"] == "Delivered",
    df_clean["Order Amount (INR)"],
    0
)
```

```python
df_clean.to_csv(
    "../data/cleaned/food_delivery_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
